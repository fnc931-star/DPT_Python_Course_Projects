### 3. Data Cleaning: (Marks 2) • Identify missing values  
• Handle missing values appropriately  
• Check and remove duplicate records if any 

```python
dataset.isna().sum()
```

```python
dataset[['MSZoning', 'Exterior1st']] = dataset[['MSZoning', 'Exterior1st']].apply(lambda x: x.fillna(x.mode()[0]))
```

```python
dataset[['TotalBsmtSF','BsmtFinSF2']] = dataset[['TotalBsmtSF','BsmtFinSF2']].fillna(dataset[['TotalBsmtSF','BsmtFinSF2']].median())
```

```python
dataset.drop(columns=['SalePrice'], inplace=True)
```

### 4. Feature Selection: (Marks 2) • Identify relevant features for analysis  
• Remove unnecessary columns such as identifiers (e.g., Id) 

```python
dataset.drop(columns=['Id'], inplace=True)
```

### 5. Data Preprocessing: (Marks 2) • Convert categorical variables into numerical form (encoding)  
• Preparing the dataset for machine learning and display it’s preview in the output 

```python
dataset.info()
```

```python
cols_to_encode = ['MSZoning', 'LotConfig', 'BldgType', 'Exterior1st']


if cols_to_encode[0] in dataset.columns:
    dataset = pd.get_dummies(dataset, columns=cols_to_encode, dtype='float64')
    print("Encoding successful!")
else:
    print("Columns already encoded or missing.")

dataset.head()
```
